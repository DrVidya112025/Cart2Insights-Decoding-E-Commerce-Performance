```python

## Business Insights & Recommendations

1. Overview

The exploratory and statistical analysis of the Cart2Insights e-commerce dataset has surfaced several significant patterns spanning sales performance,
customer experience, product categories, delivery operations, seller performance, and payment behavior.

Collectively, the findings show that delivery performance is strongly associated with customer satisfaction, that spending varies considerably across 
product categories, and that payment methods differ in their distribution across order statuses. These insights provide a foundation for identifying 
operational improvements and supporting data-driven business decisions.

2. Key Business Insights

2.1 Sales and Revenue Performance

The dataset comprises 99,441 orders, generating total revenue of approximately ₹15.84 million, based on product price and freight value combined. 
The average order value is approximately ₹160.58.

Revenue trended upward from 2017 into 2018, reflecting overall sales growth during the primary observation period. November 2017 recorded the 
highest monthly revenue, at approximately ₹11.79 lakh. Revenue for September 2018 is notably lower than surrounding months and should be interpreted 
with caution, as this appears to reflect an incomplete final reporting period rather than a genuine decline.

Business implication: Historical sales patterns can inform inventory planning, marketing campaign timing, staffing levels, and logistics capacity 
during periods of peak demand.

2.2 Product Category Spending

Revenue is distributed across a wide range of product categories, with Health & Beauty, Watches & Gifts, Bed, Bath & Table, Sports & Leisure, 
and Computers & Accessories contributing substantially to overall sales.

A One-Way ANOVA produced:

F-statistic = 153.796
p-value < 0.001

This provides statistically significant evidence that average category-level spending differs meaningfully across product categories. Descriptive 
analysis reinforces this: the Computers category recorded average order spending of approximately ₹1,286, compared with approximately ₹49 for Home 
    Comfort 2.

Business implication: Given the pronounced differences in spending behavior across categories, category-specific pricing, promotional, inventory, 
and assortment strategies are likely to be more effective than a uniform, one-size-fits-all approach.

Statistical limitation: Levene's test indicated unequal variances across categories (p < 0.001), and category sample sizes were highly unequal. 
As such, the ANOVA result, while statistically significant, should be interpreted with appropriate caution.

2.3 Delivery Performance

The average delivery time across all orders was approximately 12.50 days. Among orders with available delivery data, 89,941 were delivered on 
time or early, while 6,535 experienced delays.

A strong relationship was identified between delivery performance and customer satisfaction:

Delivery Status	Average Review Score
Early	4.29
On Time	4.03
Delayed	2.27

An independent two-sample Welch's t-test confirmed this difference is statistically significant:

t-statistic = −100.745
p-value < 0.001

Business implication: Delivery performance is a critical driver of customer experience. Proactively monitoring shipments, identifying logistics 
bottlenecks, and improving communication around delays could meaningfully reduce negative customer outcomes. It should be noted that this result reflects an association, not proof that delivery delays alone cause lower ratings.

2.4 Regional Delivery Performance

Delivery performance varies considerably by state. Average delivery times include:

SP: 8.70 days
MG: 11.95 days
PR: 11.94 days
RJ: 15.24 days
RR: 29.34 days

States with low order volumes should not be evaluated using the same lens as high-volume states. For instance, SP recorded the highest absolute count 
of delayed orders, but this is largely attributable to its substantially larger overall order volume.

Business implication: Regional logistics performance should be assessed using both average delivery time and delay rate, rather than raw delayed-order
counts alone, to avoid misattributing scale effects to genuine performance issues.

2.5 Customer Satisfaction

The overall average review score across the platform is approximately 4.09 out of 5. The distribution of review scores is as follows:

Rating	Count
5-star	57,328
4-star	19,142
3-star	8,179
2-star	3,151
1-star	11,424

Five-star reviews represent the largest segment, indicating generally favorable customer sentiment. However, the substantial gap between review scores 
for delayed versus on-time/early orders demonstrates that a strong overall average can obscure meaningful problems within specific operational segments.

Business implication: Customer experience monitoring should extend beyond the overall average rating, incorporating segmented analysis by delivery
performance, product category, region, seller, and order characteristics.

2.6 Seller Performance

Seller-level analysis reveals substantial variation in revenue, order volume, and customer ratings. Some sellers generate high revenue through 
large order volumes, while others achieve strong average ratings despite comparatively lower order counts.

Business implication: Seller performance should be evaluated holistically, using a combination of revenue, order volume, items sold, average review 
score, delivery performance, and cancellation/unavailability rate — rather than revenue in isolation.

2.7 Customer Spending Behavior

While the highest-spending customers in the dataset placed relatively large individual orders, the dataset classifies all customers as new customers. 
This reflects a structural limitation of the underlying data: the customer_id field does not reliably capture repeat purchasing behavior across orders.

Accordingly, this finding should not be interpreted as evidence that the business lacks repeat customers.

Business implication: Future retention analysis should rely on a stable, persistent customer identifier maintained across purchases. 
This would enable accurate measurement of repeat purchase rate, customer lifetime value, retention, purchase frequency, churn, and cohort behavior.

2.8 Payment Method and Order Status

A Chi-Square Test of Independence produced:

χ² = 677.083
Degrees of freedom = 28
p-value < 0.001

This provides statistical evidence that payment-record type and order status are not distributed independently. However, the assumption check 
revealed that 18 of 40 expected cells (45%) had expected frequencies below 5, meaning the standard Chi-Square approximation is not fully reliable in this context. Additionally, 2,246 orders involve more than one payment method, indicating the analysis is based on payment records rather than a strictly one-payment-method-per-order structure.

Business implication: Payment and order-status patterns merit ongoing monitoring, but this result alone should not drive operational decisions. 
Further analysis using appropriately aggregated, order-level payment data would strengthen the evidentiary basis for action.

3. Business Recommendations
Recommendation 1 — Strengthen Delivery Monitoring

Given the strong association between delayed orders and substantially lower review scores, delivery performance should be treated as a core 
customer-experience metric.

Recommended actions:

Monitor shipments as they approach their estimated delivery date.
Flag delayed orders as early as possible.
Track delivery performance by carrier and seller.
Investigate recurring regional delivery delays.
Provide customers with timely, proactive delivery-status updates.
Analyze root causes of delayed shipments.
Recommendation 2 — Develop Region-Specific Logistics Strategies

Delivery times vary considerably across regions.

Recommended actions:

Calculate delay rates, not just raw counts, at the state level.
Identify regions with consistently elevated delivery times.
Evaluate carrier performance on a regional basis.
Explore regional warehousing or fulfillment options where warranted.
Monitor regional delivery performance on an ongoing, monthly basis.

This approach helps distinguish genuine regional performance issues from effects driven simply by differences in order volume.

Recommendation 3 — Apply Category-Specific Sales Strategies

The ANOVA results confirm statistically significant differences in spending across product categories.

Recommended actions:

Develop pricing strategies tailored to individual categories.
Prioritize high-value categories for targeted promotions.
Maintain adequate inventory levels in high-revenue categories.
Evaluate pricing and assortment opportunities within lower-spending categories.
Incorporate category-level average spending into promotional planning.

Because variance is unequal across categories, decisions should also account for sample size and within-category variability.

Recommendation 4 — Prioritize Customer Experience During Delivery Delays

The gap in average review scores between delayed and on-time/early orders is substantial and warrants direct action.

Recommended actions:

Send proactive delay notifications to affected customers.
Provide accurate, updated delivery estimates.
Prioritize resolution of severely delayed shipments.
Monitor negative reviews specifically linked to delivery issues.
Investigate sellers and carriers associated with recurring delays.

The goal should be to minimize avoidable delivery problems while improving communication when delays are unavoidable.

Recommendation 5 — Establish a Seller Performance Dashboard

Seller performance should be tracked using a multi-dimensional set of KPIs:

KPI	Purpose
Seller revenue	Measures sales contribution
Order count	Measures business volume
Items sold	Measures product movement
Average review score	Measures customer satisfaction
Average delivery time	Measures fulfillment performance
Delay rate	Measures delivery reliability
Cancellation rate	Measures order reliability

This framework would help identify sellers who drive strong sales but underperform on service or delivery, as well as sellers who excel in 
customer experience.

Recommendation 6 — Improve Customer Retention Data Infrastructure

The current dataset structure prevents reliable measurement of repeat-customer behavior due to the design of the customer_id field.

For future analysis, a stable, persistent customer identifier should be maintained across all purchases. This would enable customer segmentation, 
repeat-purchase analysis, customer lifetime value calculation, retention analysis, churn prediction, and personalized marketing.

Recommendation 7 — Monitor Payment and Order Outcomes with Appropriate Caution

While the Chi-Square analysis indicates an association between payment-record type and order status, the expected-frequency assumption was violated, 
limiting the reliability of this result.

Recommended actions:

Track payment failures separately from successful transactions.
Monitor cancellation and unavailability rates by payment method.
Conduct payment-issue analysis at the order level.
Avoid drawing firm conclusions from sparsely populated payment/status categories.
Re-run the analysis after consolidating rare categories or applying a more appropriate statistical approach.

4. Priority Areas for Ongoing Business Monitoring

Based on the analysis, the following KPIs warrant regular monitoring:

Revenue and monthly sales trend
Average order value
Delivery delay rate
Average delivery time
Average review score
Review score by delivery status
Revenue by product category
Average spending by category
Seller revenue and seller rating
Regional delivery performance
Order cancellation and unavailability rates
Payment-method performance

    5. Overall Business Insight

This analysis demonstrates substantial variation in sales, spending, delivery performance, and customer satisfaction across different business 
segments. The most pronounced customer-experience finding is the relationship between delivery delays and lower review scores. Product category 
also shows statistically significant differences in spending patterns. Payment-method and order-status patterns exhibit a statistical association, 
though the limitations of the Chi-Square test require this finding to be interpreted with appropriate caution.

Overall, the analysis suggests that business performance is best monitored at the segment level, rather than through reliance on overall averages
alone. Integrating sales, category, seller, regional, delivery, customer-review, and payment metrics offers a more comprehensive foundation for 
operational decision-making.

These findings are observational in nature and describe associations present within the available dataset. They should not be interpreted as proof 
that any single factor directly causes the observed business outcomes.


```
