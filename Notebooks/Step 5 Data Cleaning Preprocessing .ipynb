## Create cleaned copies and Remove exact duplicate rows

```python

# Create cleaning copies
customers_clean = customers.copy()
orders_clean = orders.copy()
order_items_clean = order_items.copy()
order_payments_clean = order_payments.copy()
order_reviews_clean = order_reviews.copy()
products_clean = products.copy()
sellers_clean = sellers.copy()
geolocation_clean = geolocation.copy()
category_translation_clean = category_translation.copy()


# Store original row counts
before_counts = {
    "customers": len(customers_clean),
    "orders": len(orders_clean),
    "order_items": len(order_items_clean),
    "order_payments": len(order_payments_clean),
    "order_reviews": len(order_reviews_clean),
    "products": len(products_clean),
    "sellers": len(sellers_clean),
    "geolocation": len(geolocation_clean),
    "category_translation": len(category_translation_clean)
}


# Remove exact duplicate rows
customers_clean = customers_clean.drop_duplicates().reset_index(drop=True)
orders_clean = orders_clean.drop_duplicates().reset_index(drop=True)
order_items_clean = order_items_clean.drop_duplicates().reset_index(drop=True)
order_payments_clean = order_payments_clean.drop_duplicates().reset_index(drop=True)
order_reviews_clean = order_reviews_clean.drop_duplicates().reset_index(drop=True)
products_clean = products_clean.drop_duplicates().reset_index(drop=True)
sellers_clean = sellers_clean.drop_duplicates().reset_index(drop=True)
geolocation_clean = geolocation_clean.drop_duplicates().reset_index(drop=True)
category_translation_clean = category_translation_clean.drop_duplicates().reset_index(drop=True)


# Create clean dataset dictionary
clean_datasets = {
    "customers": customers_clean,
    "orders": orders_clean,
    "order_items": order_items_clean,
    "order_payments": order_payments_clean,
    "order_reviews": order_reviews_clean,
    "products": products_clean,
    "sellers": sellers_clean,
    "geolocation": geolocation_clean,
    "category_translation": category_translation_clean
}


# Display results
print("=" * 70)
print("DUPLICATE REMOVAL SUMMARY")
print("=" * 70)

for name, df in clean_datasets.items():
    removed = before_counts[name] - len(df)

    print(
        f"{name:25} "
        f"Before: {before_counts[name]:8} | "
        f"After: {len(df):8} | "
        f"Removed: {removed}"
    )
```

# OUTLIER DETECTION USING IQR

```python

def check_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(f"\n{column}")
    print("-" * 50)
    print("Q1:", round(Q1, 2))
    print("Q3:", round(Q3, 2))
    print("Lower limit:", round(lower, 2))
    print("Upper limit:", round(upper, 2))
    print("Outlier count:", len(outliers))


# ORDER ITEMS
check_outliers(order_items_clean, "price")
check_outliers(order_items_clean, "freight_value")

# PAYMENTS
check_outliers(order_payments_clean, "payment_value")
check_outliers(order_payments_clean, "payment_installments")

# PRODUCTS
check_outliers(products_clean, "product_weight_g")
check_outliers(products_clean, "product_length_cm")
check_outliers(products_clean, "product_height_cm")
check_outliers(products_clean, "product_width_cm")
```

```python
# ============================================================
# OUTLIER FLAGS
# ============================================================

def add_outlier_flag(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[column + "_outlier_flag"] = (
        (df[column] < lower) |
        (df[column] > upper)
    )


# ORDER ITEMS
add_outlier_flag(order_items_clean, "price")
add_outlier_flag(order_items_clean, "freight_value")

# PAYMENTS
add_outlier_flag(order_payments_clean, "payment_value")
add_outlier_flag(order_payments_clean, "payment_installments")

# PRODUCTS
add_outlier_flag(products_clean, "product_weight_g")
add_outlier_flag(products_clean, "product_length_cm")
add_outlier_flag(products_clean, "product_height_cm")
add_outlier_flag(products_clean, "product_width_cm")


print("Outlier flags created successfully.")
```
