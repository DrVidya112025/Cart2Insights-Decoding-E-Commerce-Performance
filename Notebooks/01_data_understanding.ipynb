## Dataset Overview

| Table | Grain | Purpose |
|---|---|---|
| customers | Customer | Customer information |
| orders | Order | Order and delivery lifecycle |
| order_items | Order item | Products purchased and seller/price details |
| order_payments | Payment | Payment information |
| order_reviews | Review | Customer ratings and review information |
| products | Product | Product attributes and categories |
| sellers | Seller | Seller information |
| geolocation | Geolocation record | Location information |
| product_category_name_translation | Category | Category name translation |

## Key Relationships

- customers -> orders : 1-to-many
- orders -> order_items : 1-to-many
- orders -> order_payments : 1-to-many
- orders -> order_reviews : 1-to-many (maybe)
- products -> order_items : 1-to-many
- sellers -> order_items : 1-to-many
- product_category_name_translation -> products : 1-to-many
- customer/seller ZIP prefix -> geolocation

## Keys

| Table | Primary Key | Foreign Keys |
|---|---|---|
| customers | customer_id | — |
| orders | order_id | customer_id |
| order_items | (order_id, order_item_id) | order_id, product_id, seller_id |
| order_payments | (order_id, payment_sequential) | order_id |
| order_reviews | review_id* | order_id |
| products | product_id | product_category_name |
| sellers | seller_id | seller_zip_code_prefix |
| geolocation | geolocation_zip_code_prefix* | — |
| category_translation | product_category_name | — |

## Data Grain

| Table | One row represents |
|---|---|
| customers | One customer record |
| orders | One order |
| order_items | One item within an order |
| order_payments | One payment record within an order |
| order_reviews | One review record |
| products | One product |
| sellers | One seller |
| geolocation | One geolocation record |
| category_translation | One category translation |

## Cardinality

The dataset contains several one-to-many relationships.

For example:

1 customer → many orders
1 order → many order items
1 order → many payment records
1 seller → many order items
1 product → many order items

Therefore, joins must be performed with careful consideration of the
resulting grain.
