```python
##4.Number of orders placed by the customer 
query = """
SELECT
    customer_id,
    COUNT(DISTINCT order_id) AS customer_order_count
FROM orders
GROUP BY customer_id;
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
```

```python
## 5.Total amount spent by the customer 
query = """
SELECT
    o.customer_id,
    ROUND(SUM(oi.price + oi.freight_value), 2) AS customer_total_spending
FROM orders o
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY o.customer_id;
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
```

```python
## 6. Average spending per order
query = """
SELECT
    o.customer_id,
    ROUND(
        SUM(oi.price + oi.freight_value) / COUNT(DISTINCT o.order_id),
        2
    ) AS average_order_value
FROM orders o
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY o.customer_id;
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
```

```python
## 7. Total revenue generated by the seller  
query = """
SELECT
    seller_id,
    ROUND(SUM(price), 2) AS seller_revenue
FROM order_items
GROUP BY seller_id;
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
```

```python
## 8.Number of orders handled by the seller
query = """
SELECT
    seller_id,
    COUNT(DISTINCT order_id) AS seller_order_count
FROM order_items
GROUP BY seller_id;
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
```

```python
## 9.Repeat customer 1 if customer has more than one order, otherwise 0 
query = """
SELECT
    customer_id,
    CASE
        WHEN COUNT(DISTINCT order_id) > 1 THEN 1
        ELSE 0
    END AS repeat_customer
FROM orders
GROUP BY customer_id;
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
```

More than 1 order → 1 (Repeat Customer)
1 order           → 0 (Not Repeat Customer)
